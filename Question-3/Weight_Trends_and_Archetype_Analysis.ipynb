### Define the archetype classification function
Defines the row-wise rules that assign each athlete-event record to one of the seven functional archetypes. Athletics is classified using event keywords; Swimming uses event distance; Cycling distinguishes sprint from other cycling events; remaining sports use `sport_mapping`.

```python
import re

def classify_archetype(row):

    sport = row["Sport"]
    event = str(row["Event"]).lower()

    # Athletics
    if sport == "Athletics":

        if "decathlon" in event or "heptathlon" in event:
            return "Technical / Mixed"

        endurance_keywords = [
            "marathon", "kilometres walk",
            "10,000 metres", "5,000 metres",
            "3,000 metres", "1,500 metres",
            "800 metres"
        ]

        if any(word in event for word in endurance_keywords):
            return "Endurance"

        return "Speed & Power"

    # Swimming
    elif sport == "Swimming":

        distance = int(re.search(r"\d+", event).group())

        if distance >= 400:
            return "Endurance"
        else:
            return "Speed & Power"

    # Cycling
    elif sport == "Cycling":

        if "sprint" in event:
            return "Speed & Power"
        else:
            return "Endurance"

    # All other sports
    else:
        return sport_mapping.get(sport, pd.NA)
```

### Apply archetype classification
Runs `classify_archetype` on every row and stores the result in the `Archetype` column. Each row is classified using both its sport and, where needed, its event.

```python
df["Archetype"] = df.apply(classify_archetype, axis=1)
```

### Define the descriptive-statistics function
Defines a reusable function that groups a supplied variable by archetype and calculates count, mean, median, standard deviation, minimum, maximum, first quartile, third quartile and interquartile range (Q3 minus Q1).

```python
def descriptive_statistics(data, column):

    stats = data.groupby("Archetype")[column].agg(
        Count="count",
        Mean="mean",
        Median="median",
        Std_Dev="std",
        Minimum="min",
        Maximum="max",
        Q1=lambda x: x.quantile(0.25),
        Q3=lambda x: x.quantile(0.75)
    )

    stats["IQR"] = stats["Q3"] - stats["Q1"]

    return stats.reset_index()
```

### Create the common complete-case sample for Model B
Excludes rows missing any of ID, Year, Height, Weight, Sex or Archetype. It also reports the number of observations and the distribution of archetypes and sex in the resulting sample.

### Refit Model A on the common sample
Refits the height-and-year model using `model_data`, the same complete-case rows that Model B will use. This is important for a fair nested-model comparison because the two models then differ in predictors, not in the observations included.

```python
model_A_common = smf.ols(
    "Weight ~ Height + Year_Decade",
    data=model_data
).fit(
    cov_type="cluster",
    cov_kwds={"groups": model_data["ID"]}
)
```

### Fit Model B with archetype and sex controls
Fits weight as a function of height, decade-scaled year, categorical archetype and categorical sex. The `C()` notation tells statsmodels to treat archetype and sex as categories rather than numeric quantities. Athlete-ID clustered standard errors account for repeated records within athletes.

```python
model_B = smf.ols(
    "Weight ~ Height + Year_Decade + C(Archetype) + C(Sex)",
    data=model_data
).fit(
    cov_type="cluster",
    cov_kwds={"groups": model_data["ID"]}
)

print(model_B.summary())
```

### Extract Model B's key results
Prints the adjusted year coefficient, p-value, 95% confidence interval, R-squared and number of observations. These are the main quantities used to interpret the composition-adjusted model.

```python
print("MODEL B RESULTS")

print("Year coefficient:",
      model_B.params["Year_Decade"])

print("P-value:",
      model_B.pvalues["Year_Decade"])

print("95% Confidence interval:")
print(model_B.conf_int().loc["Year_Decade"])

print("R-squared:", model_B.rsquared)

print("Observations:", int(model_B.nobs))
```

### Compare the two models on the same sample
