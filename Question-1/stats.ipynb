```python
import pandas as pd
import numpy as np
df = pd.read_csv("q1_working.csv")
```

```python
metrics = ['Age','Height','Weight','BMI']
print(df.shape)
```

```python
## checking archetype sample sizes
counts = df.groupby("Archetype").agg(
    observations=("ID", "size"),
    unique_athletes=("ID", "nunique")
).reset_index()

display(counts)

counts.to_csv("q1_archetype_counts.csv", index=False)
```

```python
## overall stats by each archtype
stats = (
    df.groupby("Archetype")[metrics]
    .agg(["count", "mean", "median", "std",
          "min", "max",
          lambda x: x.quantile(0.25),
          lambda x: x.quantile(0.75)])
)

# Renaming  lambda functions
stats.columns = pd.MultiIndex.from_tuples([
    (metric, "Q1" if stat == "<lambda_0>" else
             "Q3" if stat == "<lambda_1>" else stat)
    for metric, stat in stats.columns
])

# Calculating  IQR
for metric in metrics:
    stats[(metric, "IQR")] = (
        stats[(metric, "Q3")] - stats[(metric, "Q1")]
    )

display(stats.round(2))

stats.to_csv("q1_overall_statistics.csv")
```

```python
## gender wise stats

sex_stats = (
    df.groupby(["Archetype", "Sex"])[metrics]
    .agg(["count", "mean", "median", "std",
          "min", "max",
          lambda x: x.quantile(0.25),
          lambda x: x.quantile(0.75)])
)

sex_stats.columns = pd.MultiIndex.from_tuples([
    (metric, "Q1" if stat == "<lambda_0>" else
             "Q3" if stat == "<lambda_1>" else stat)
    for metric, stat in sex_stats.columns
])

for metric in metrics:
    sex_stats[(metric, "IQR")] = (
        sex_stats[(metric, "Q3")] -
        sex_stats[(metric, "Q1")]
    )

display(sex_stats.round(2))

sex_stats.to_csv("q1_sexwise_statistics.csv")
```

```python
## mean comparison

mean_comparison = (
    df.groupby("Archetype")[metrics]
    .mean()
    .round(3)
)

display(mean_comparison)

mean_comparison.to_csv("q1_mean_comparison.csv")
```

```python
## highest and lowest
for metric in metrics:
    highest = mean_comparison[metric].idxmax()
    lowest = mean_comparison[metric].idxmin()

    print(f"\n{metric}")
    print(f"Highest mean: {highest} ({mean_comparison.loc[highest, metric]})")
    print(f"Lowest mean: {lowest} ({mean_comparison.loc[lowest, metric]})")
```
