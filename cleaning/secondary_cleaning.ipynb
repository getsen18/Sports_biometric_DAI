```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('final_data.csv')
```

### We are replacing the missing ages with median rather than deleting them completely

```python
df['Age'] = df.groupby(['Sport', 'Sex'])['Age'].transform(
    lambda x: x.fillna(x.median())
).fillna(df['Age'].median())

df
```

## filling missing medal values with string 'no medal'

```python
df['Medal'] = df['Medal'].fillna('No Medal')
```

```python
df['Medal']
```

#### stripping the extra sport names

```python

df['Event'] = df.apply(
    lambda row: row['Event'][len(row['Sport']):].strip() 
    if row['Event'].startswith(row['Sport']) else row['Event'], 
    axis=1
)

df['Event']
```

# removing dupilicate team names

```python
df[df['Team']!="India"]
```

```python
df['Team'] = 'India'
```

```python
df[df['Team']!="India"]
```

### dropping zero variance coloumns here the country / team always has india drop them

```python
df = df.drop(
    columns=['index', 'NOC', 'Season', 'Games', 'Team']
)
```

```python
df
```

```python
## Cleaning athelete names i.e removing (nicknames) and " "(quoted ones)
```

```python
df['Clean_Name'] = (
    df['Name']
    .str.replace(r'"[^"]*"', '', regex=True)
    .str.replace(r'\([^)]*\)', '', regex=True)
    .str.replace(r'\s+', ' ', regex=True)
    .str.strip()
)
```
