```python
import numpy as np
import pandas as pd
import seaborn as sbn
```

```python
df = pd.read_csv('data.csv')
```

```python
# Filtering data set to only include summer and Indian athletes
newdf = df[(df["Season"]=="Summer")&(df["NOC"]=="IND")]
newdf
```

```python
## droping duplicates
newdf = newdf.drop_duplicates()
newdf
```

```python
## dropping nan in height or weight
newdf = newdf.dropna(subset=["Height","Weight"])
newdf
```

```python
# sorting the final data 
newdf = newdf.sort_values(by=["ID","Year"]).reset_index()
newdf
```

```python
newdf.to_csv("final_data.csv",index=False)
```
